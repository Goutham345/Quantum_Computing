```python
!pip install qiskit
!pip install qiskit-aer
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
Imsg = "11"
qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
if Imsg[1] == "1":
    qc.x(0)
if Imsg[0] == "1":
    qc.z(0)
qc.cx(0, 1)
qc.h(0)
qc.measure([0, 1], [1, 0])
backend = AerSimulator()
counts = backend.run(
    transpile(qc, backend),
    shots=1024
).result().get_counts()
print(counts)
```
